#### クラスタ分析
ラベルなしデータの分析

k-means法を使った類似度のよるオブジェクトのグループ化


```python
from sklearn.datasets import make_blobs
import numpy as np
n_centers=4
n_features=2
n_samples=100
#X: 特徴量ベクトル
#y: クラスタ(正解)
X, y = make_blobs(
  n_samples = n_samples,#NOTE: データ点の総数
  n_features = n_features, #NOTE: 特徴量の個数,
  centers = n_centers, #クラスタの個数,
  cluster_std = 0.9, #クラスタ内の標準偏差,
  shuffle = True, #データ点のシャッフル
  random_state = 0#乱数発生機の状態
)

import matplotlib.pyplot as plt
import seaborn as sns; sns.set()



from kmeansSelf import kMeans



model = kMeans(X, n_centers = 4)

centroids = []
mse = []
for _ in range(10):
  model.fit1step()
  centroids.append(model._centroids)
  mse.append(model._mse)

colors = ['green','red','blue','black']


for idx, color in enumerate(colors):
  plt.scatter(X[model.results==idx,0],X[model.results==idx,1], c = color, marker = 'o', s = 20, alpha = 0.3)
  ci = np.array(centroids)[:,idx,:].T
  plt.scatter(ci[0],ci[1], c = color, marker = '+', s = 50, alpha = 1)
  plt.plot(ci[0],ci[1], c = color, linestyle = 'solid')


plt.grid()


plt.tight_layout()
plt.show()

```
